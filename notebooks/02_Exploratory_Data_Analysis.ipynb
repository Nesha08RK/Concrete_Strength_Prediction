```python
# Import lbraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
# Load Dataset
df = pd.read_csv("../Dataset.csv")

df.head()

# Check Dataset Shape
print("Dataset Shape:", df.shape)

# Check Duplicate Records
print("Duplicate Records:", df.duplicated().sum())
df = df.drop_duplicates() #if duplicates are found, drop them
```

```python
# Check Missing Values
df.isnull().sum()

# Statistical Summary
df.describe()

# Plot Histograms
df.hist(figsize=(15,10))

plt.tight_layout()

plt.show()
```

```python
# Correlation Matrix
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

```python
# Boxplots
plt.figure(figsize=(15,8))

df.boxplot(rot=90)

plt.show()
```

```python
# Distribution of Target Variable
plt.figure(figsize=(6,4))

sns.histplot(df["Strength"], bins=30, kde=True)

plt.title("Distribution of Strength")

plt.show()
```

```python
# Scatter Plots

# Cement vs Strength
sns.scatterplot(x=df["Cement"], y=df["Strength"])
plt.title("Cement vs Strength")
plt.show()

# Water vs Strength
sns.scatterplot(x=df["Water"], y=df["Strength"])
plt.title("Water vs Strength")
plt.show()

# Age vs Strength
sns.scatterplot(x=df["Age"], y=df["Strength"])
plt.title("Age vs Strength")
plt.show()

# Fly Ash vs Strength
sns.scatterplot(x=df["Fly Ash"], y=df["Strength"])
plt.title("Fly Ash vs Strength")
plt.show()
```

Observations

• The dataset contains 1030 samples and 9 features.

• A total of 25 duplicate records were identified in the dataset.

• No missing values were found.

• Cement shows the highest positive correlation (0.49) with compressive strength.

• Superplasticizer and Age also have a moderate positive correlation with strength.

• Water has a negative correlation (-0.27), indicating that increasing water content generally reduces concrete strength.

• The visualizations help understand the distribution of features and their relationship with the target variable.

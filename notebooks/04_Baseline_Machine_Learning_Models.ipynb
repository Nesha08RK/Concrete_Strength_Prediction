Objective

Train and compare three basic machine learning models:

Linear Regression
Decision Tree Regressor
Random Forest Regressor

using the processed concrete dataset.


```python
# Import Libraries
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
```

```python
# Load Dataset

X_train = pd.read_csv("../dataset/X_train.csv")
X_test = pd.read_csv("../dataset/X_test.csv")

y_train = pd.read_csv("../dataset/y_train.csv")
y_test = pd.read_csv("../dataset/y_test.csv")
```

```python
# Check Shapes
print(X_train.shape)
print(X_test.shape)

print(y_train.shape)
print(y_test.shape)
```

```python
# Train Linear Regression
lr = LinearRegression()
lr.fit(X_train, y_train)
```

```python
# Prediction
lr_prediction = lr.predict(X_test)

# Evaluate Linear Regression
lr_mae = mean_absolute_error(y_test, lr_prediction)

lr_mse = mean_squared_error(y_test, lr_prediction)

lr_rmse = np.sqrt(lr_mse)

lr_r2 = r2_score(y_test, lr_prediction)

print("Linear Regression")

print("MAE :", lr_mae)

print("MSE :", lr_mse)

print("RMSE :", lr_rmse)

print("R2 Score :", lr_r2)
```

```python
# Train Decision Tree
dt = DecisionTreeRegressor(random_state=42)
dt.fit(X_train, y_train)
```

```python
# Prediction
dt_prediction = dt.predict(X_test)

# Evaluate Decision Tree
dt_mae = mean_absolute_error(y_test, dt_prediction)

dt_mse = mean_squared_error(y_test, dt_prediction)

dt_rmse = np.sqrt(dt_mse)

dt_r2 = r2_score(y_test, dt_prediction)

print("Decision Tree")

print("MAE :", dt_mae)

print("MSE :", dt_mse)

print("RMSE :", dt_rmse)

print("R2 Score :", dt_r2)
```

```python
# Train Random Forest
rf = RandomForestRegressor(
    random_state=42
)

rf.fit(X_train, y_train)
```

```python
# Prediction
rf_prediction = rf.predict(X_test)

# Evaluate Random Forest
rf_mae = mean_absolute_error(y_test, rf_prediction)

rf_mse = mean_squared_error(y_test, rf_prediction)

rf_rmse = np.sqrt(rf_mse)

rf_r2 = r2_score(y_test, rf_prediction)

print("Random Forest")

print("MAE :", rf_mae)

print("MSE :", rf_mse)

print("RMSE :", rf_rmse)

print("R2 Score :", rf_r2)
```

```python
# Compare Models
results = pd.DataFrame({

    "Model":[
        "Linear Regression",
        "Decision Tree",
        "Random Forest"
    ],

    "MAE":[
        lr_mae,
        dt_mae,
        rf_mae
    ],

    "MSE":[
        lr_mse,
        dt_mse,
        rf_mse
    ],

    "RMSE":[
        lr_rmse,
        dt_rmse,
        rf_rmse
    ],

    "R2 Score":[
        lr_r2,
        dt_r2,
        rf_r2
    ]

})

results
```

```python
# Best Model
best_model = results.loc[
    results["R2 Score"].idxmax()
]

print(best_model)
```

## Observations

- Three baseline machine learning models, namely **Linear Regression**, **Decision Tree Regressor**, and **Random Forest Regressor**, were successfully trained and evaluated using the concrete compressive strength dataset.

- **Linear Regression** achieved an **R² Score of 0.5802**, indicating that it was less effective in capturing the complex relationship between the input features and concrete compressive strength.

- **Decision Tree Regressor** improved the prediction performance by achieving an **R² Score of 0.8824** with lower prediction errors than Linear Regression.

- **Random Forest Regressor** achieved the highest prediction accuracy among the baseline models with an **R² Score of 0.9074**, **MAE of 3.4979**, and **RMSE of 5.2552**.

- The comparison demonstrates that **ensemble learning methods**, particularly **Random Forest**, provide better prediction performance than conventional regression techniques for concrete compressive strength prediction.

- Based on the evaluation metrics, **Random Forest Regressor** is selected as the **best baseline model** and will be compared with advanced machine learning algorithms such as **XGBoost**, **CatBoost**, and **LightGBM** in the next phase of the project.
